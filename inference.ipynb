# Inference

Inisiasi pustaka:

```python
import joblib
import tensorflow as tf
import numpy as np
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

Inference:

```python
model = tf.keras.models.load_model('model.h5')
tokenizer = joblib.load('tokenizer.pkl')
```

```python
def predict_text(input_text, model, tokenizer, max_sequence_length=100):
    sequences = tokenizer.texts_to_sequences([input_text])
    X_input = pad_sequences(sequences, maxlen=max_sequence_length)
    
    prediction = model.predict(X_input)
    
    predicted_class = np.argmax(prediction, axis=1)
    return predicted_class
```

```python
input_text = "Livin' saat ini susah digunakan"

predicted_class = predict_text(input_text, model, tokenizer)
class_names = ['negatif', 'netral', 'positif']

print("===== INFERENSI =====")
print(f"Ulasan: {input_text}")
print(f"Prediksi kelas untuk teks: {class_names[predicted_class[0]]}")
```
